```python
#Ejercicio 3.9.1. Escribir dos funciones que permitan calcular:
#a) La duración en segundos de un intervalo dado en horas, minutos y segundos.

def duracion_segundos(horas, minutos, segundos):
    return horas * 3600 + minutos * 60 + segundos

#b) La duración en horas, minutos y segundos de un intervalo dado en segundos.

def duracion_horas_minutos_segundos(segundos):
    horas = segundos // 3600
    segundos = segundos % 3600
    minutos = segundos // 60
    segundos = segundos % 60
    return horas, minutos, segundos
```

```python
#Ejercicio 3.9.2. Usando las funciones del ejercicio anterior, escribir un programa que pida al
#usuario dos intervalos expresados en horas, minutos y segundos, sume sus duraciones, ç
#y muestre por pantalla la duración total en horas, minutos y segundos.

horas1 = float(input("Ingrese las horas del primer intervalo: "))
minutos1 = float(input("Ingrese los minutos del primer intervalo: "))
segundos1 = float(input("Ingrese los segundos del primer intervalo: "))

horas2 = float(input("Ingrese las horas del segundo intervalo: "))
minutos2 = float(input("Ingrese los minutos del segundo intervalo: "))
segundos2 = float(input("Ingrese los segundos del segundo intervalo: "))

# Convertir ambos intervalos a segundos
total_segundos = duracion_segundos(horas1, minutos1, segundos1) + duracion_segundos(horas2, minutos2, segundos2)

# Convertir la duración total a horas, minutos y segundos
horas_total, minutos_total, segundos_total = duracion_horas_minutos_segundos(total_segundos)

print(f"La duración total es: {horas_total} horas, {minutos_total} minutos y {segundos_total} segundos")
```

```python
#Ejercicio 3.9.3. Escribir una función que, dados cuatro números, devuelva el mayor producto
#de dos de ellos. Por ejemplo, si recibe los números 1, 5, -2, -4 debe devolver 8, que es el producto
#más grande que se puede obtener entre ellos (8 = −2 × −4).

def mayor_producto(n1, n2, n3, n4):
    productos = [
        n1 * n2,
        n1 * n3,
        n1 * n4,
        n2 * n3,
        n2 * n4,
        n3 * n4
    ]
    return max(productos)
```

```python
#Ejercicio 3.9.4. Área de polígonos
#a) Escribir una función que reciba las coordenadas de un vector en ℝ3
#(x,y,z) y devuelva
#la norma del vector, dada por ‖(𝑥, 𝑦, 𝑧)‖ = √𝑥2 + 𝑦2 + 𝑧2.
#Ejemplo: norma(3, 2, -4) → 5.3851

def norma(x, y, z):
    return (x**2 + y**2 + z**2) ** 0.5

#b) Escribir una función que reciba las coordenadas de dos vectores en ℝ3
#(x1,y1,z1,x2,
#y2,z2) y devuelva las coordenadas del vector diferencia (debe devolver 3 valores numéricos).
#Ejemplo: diferencia(8, 7, -3, 5, 3, 2) → (3, 4, -5)

def diferencia(x1, y1, z1, x2, y2, z2):
    return (x1 - x2, y1 - y2, z1 - z2)

#c) Escribir una función que reciba las coordenadas de dos vectores en ℝ3 y devuelva las
#coordenadas del producto vectorial, definido como:
#(𝑥1, 𝑦1, 𝑧1) * (𝑥2, 𝑦2, 𝑧2) =  (𝑦1𝑧2 − 𝑧1𝑦2, 𝑧1𝑥2 − 𝑥1𝑧2, 𝑥1𝑦2 − 𝑦1𝑥2)
#Ejemplo: producto_vec(1, 4, -2, 3, -1, 0) → (-2, -6, -13)

def producto_vec(x1, y1, z1, x2, y2, z2):
    return (y1 * z2 - z1 * y2, z1 * x2 - x1 * z2, x1 * y2 - y1 * x2)

#d) Utilizando las funciones anteriores, escribir una función que reciba las coordenadas de
#3 puntos en ℝ3 y devuelva el área del triángulo que conforman.
#Ayuda: Si 𝐴, 𝐵 y 𝐶 son 3 puntos en el espacio, la norma del producto vectorial es
#igual al doble del área del triángulo 𝐴𝐵𝐶.
#Ejemplo: area_triangulo(5, 8, -1, -2, 3, 4, -3, 3, 0) → 19.4551

def area_triangulo(x1, y1, z1, x2, y2, z2, x3, y3, z3):
    # Calcular los vectores AB y AC
    AB = diferencia(x2, y2, z2, x1, y1, z1)
    AC = diferencia(x3, y3, z3, x1, y1, z1)
    
    # Calcular el producto vectorial de AB y AC
    producto = producto_vec(AB[0], AB[1], AB[2], AC[0], AC[1], AC[2])
    
    # Calcular la norma del producto vectorial
    norma_producto = norma(producto[0], producto[1], producto[2])
    
    # El área del triángulo es la mitad de la norma del producto vectorial
    return norma_producto / 2

#e) Escribir una función que reciba las coordenadas de 4 puntos en el plano ℝ2
#(x1,y1,x2,
#y2,x3,y3,x4,y4) que conforman un cuadrilátero convexo, y devuelva el área del mismo.
#Ayuda: Aprovechar las funciones escritas anteriormente, asumiendo que los puntos dados
#están en ℝ3
#con coordenada 𝑧 = 0.
#Ejemplo: area_cuadrilatero(4, 3, 5, 10, -2, 8, -3, -5) → 65

def area_cuadrilatero(x1, y1, x2, y2, x3, y3, x4, y4):
    # Calcular el área de los dos triángulos que forman el cuadrilátero
    area1 = area_triangulo(x1, y1, 0, x2, y2, 0, x3, y3, 0)
    area2 = area_triangulo(x1, y1, 0, x3, y3, 0, x4, y4, 0)
    
    # El área del cuadrilátero es la suma de las áreas de los dos triángulos
    return area1 + area2
```
