```python
#10. Calculate the factorial of a given number

my_number = int(input("Write a number: "))
factorial = 1
for i in range(2, my_number + 1):
	factorial = i * factorial
print("The factorial of " + str(my_number) + " is " + str(factorial))
```

```python
#11. Calculate the n-th term of the Fibonacci succession

my_number = int(input("Write a number: "))
if my_number < 2:
	result = my_number
else:
	one_step_before = 1
	two_steps_before = 2
	for i in range(1, my_number):
		result = one_step_before + two_steps_before
		two_steps_before = one_step_before
		one_step_before = result
print("The n-th term of the Fibonacci succession is " + str(result))
```

```python
#16. Determine whether a given number is prime

my_number = int(input("Write a number here: "))
is_prime = True
if my_number <= 1:
    is_prime = False
else:
    for i in range(2, int(my_number ** 0.5) + 1):
        if my_number % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{my_number} is a prime number.")
else:
    print(f"{my_number} is not a prime number.")
```
