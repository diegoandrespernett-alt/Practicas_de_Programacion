```python
#11. Calculate the n-th term of the Fibonacci succession

my_number = int(input("Write a number: "))
if my_number < 2:
	result = my_number
else:
	one_step_before = 1
	two_steps_before = 2
	for i in range(1, my_number):
		result = one_step_before + two_steps_before
		two_steps_before = one_step_before
		one_step_before = result
print("The n-th term of the Fibonacci succession is " + str(result))
```

```python
#16. Determine whether a given number is prime

my_number = int(input("Write a number here: "))
is_prime = True
if my_number <= 1:
    is_prime = False
else:
    for i in range(2, int(my_number ** 0.5) + 1):
        if my_number % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{my_number} is a prime number.")
else:
    print(f"{my_number} is not a prime number.")
```

```python
#17. Determine whether a given number is a power of two (or any other number)

my_number = int(input("Write a number here: "))
base = int(input("Write the base here: "))

if my_number <= 0 or base <= 1:
    print("The number must be positive and the base must be greater than 1.")
else:
    # Check if my_number is a power of base
    temp = my_number
    while temp > 1:
        if temp % base != 0:
            print(f"{my_number} is not a power of {base}.")
            break
        temp //= base
    else:
        print(f"{my_number} is a power of {base}.")
```

```python
#20. Solve a second degree equation, given the values of a, b and c.

a = float(input("Write the coefficient a here: "))
b = float(input("Write the coefficient b here: "))
c = float(input("Write the coefficient c here: "))

delta = b**2 - 4*a*c
if delta < 0:
    print("The equation has no real solutions.")
if delta == 0:
    x1 = -b / (2*a)
    x2 = x1
    print(f"The equation has one real solution: x = {x1}")
else:
    x1 = (-b + delta**0.5) / (2*a)
    x2 = (-b - delta**0.5) / (2*a)
    print(f"The equation has two real solutions: x1 = {x1}, x2 = {x2}")
```

```python
#22. Calculate the n-th term of the seccession: an = (-1)^n * (n^2-1)/2n.

from fractions import Fraction

def nth_term(n):
    if n < 1:
        raise ValueError("n debe ser un entero positivo")
    return (-1) ** n * (n ** 2 - 1) / (2 * n)

# Versión exacta con fracciones (evita errores de decimales)
def nth_term_exact(n):
    if n < 1:
        raise ValueError("n debe ser un entero positivo")
    return Fraction((-1) ** n * (n ** 2 - 1), 2 * n)


n = 5
print(f"a_{n} =", nth_term(n))
print(f"a_{n} (exacto) =", nth_term_exact(n))

# Primeros términos de la sucesión
print("\nPrimeros 5 términos:")
for i in range(1, 6):
    print(f"a_{i} = {nth_term(i)}  ({nth_term_exact(i)})")
```
