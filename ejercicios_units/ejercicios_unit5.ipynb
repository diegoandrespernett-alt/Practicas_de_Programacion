```python
#Solve the following exercises using recursion. Also implement the iterative solution, if it has not been provided previously:

#1) Write a function to calculate the power n of a certain number.

def power(base, exponent):
    if exponent == 0:
        return 1
    else:
        return base * power(base, exponent - 1)
```

```python
#2) Write a function to represent a given base 10 number in another base B (with B < 10).

def convert_to_base(number, base):
    if number < base:
        return str(number)
    else:
        return convert_to_base(number // base, base) + str(number % base)

number = 255
base = 7
result = convert_to_base(number, base)
print(f"The representation of {number} in base {base} is: {result}")
```

```python
#3) Write a function to calculate the multiplication of two numbers recursively.

def multiply(a, b):
    if b == 0:
        return 0
    elif b < 0:
        return -multiply(a, -b)
    else:
        return a + multiply(a, b - 1)

a = 5
b = 3
result = multiply(a, b)
print(f"The multiplication of {a} and {b} is: {result}")
```

```python
#4) Write a function that checks if the digits of a given number are written in ascending or descending order. The function must make use of a key argument that takes a default value to determine which order must verify (for example, check ascending order by default).

def check_order(number, ascending=True):
    digits = [int(d) for d in str(number)]
    if ascending:
        return all(digits[i] <= digits[i + 1] for i in range(len(digits) - 1))
    else:
        return all(digits[i] >= digits[i + 1] for i in range(len(digits) - 1))

input_number = 12345
is_ascending = check_order(input_number)
print(f"The digits of {input_number} are in ascending order: {is_ascending}")
```

```python
#5) Write a recursive function gcd(a, b) that finds the largest positive integer that divides both given numbers. Use Euclid's algorithm: the greatest common divisor of a and b is the same as the greatest common divisor of b and the remainder of the division a/b. When b reaches 0, the answer is a.

def gcd(a, b):
    if b == 0:
        return a
    else:
        return gcd(b, a % b)

a = 48
b = 18
result = gcd(a, b)
print(f"The greatest common divisor of {a} and {b} is: {result}")
```

```python
#6) Implement the calculation of a combinatorial number:
#    (m, n) = (m-1, n-1) + (m-1, n)

#The value (m, n) is 1 when:
#· m = 0
#· n = 1
#· m = n

def combinatorial(m, n):
    if m == 0 or n == 1 or m == n:
        return 1
    else:
        return combinatorial(m - 1, n - 1) + combinatorial(m - 1, n)

m = 5
n = 3
result = combinatorial(m, n)
print(f"The result of the given matrix (m, n) = ({m}, {n}) is: {result}")
```
