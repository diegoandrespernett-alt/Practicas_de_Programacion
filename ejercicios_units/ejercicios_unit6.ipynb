```python
# Linear search Case

input_list = ["apple", "banana", "cherry", "date", "fig", "grape"]

def linear_search(input_list, target):
    for index in range(len(input_list)):
        if input_list[index] == target:
            return index
    return -1

# Test the function
target = "cherry"
result = linear_search(input_list, target)
print(f"Index of {target}: {result}")
```

```python
# Binary search Case

def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        # Find the middle index
        mid = left + (right - left) // 2

        # Check if target is present at mid
        if arr[mid] == target:
            return mid
        # If target is greater, ignore left half
        elif arr[mid] < target:
            left = mid + 1
        # If target is smaller, ignore right half
        else:
            right = mid - 1

    # Target was not found in the array
    return -1

# Example usage
numbers = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
result = binary_search(numbers, 23) # Returns index 5
print(f"Index is {result}")
```

```python
# Bubble sorting Case

def bubble_sort(list):
    n = len(list)

    for i in range(n):
        for j in range(0, n - i - 1):

            if list[j] > list[j + 1]:
                list[j], list[j + 1] = list[j + 1], list[j]

    return list


numbers = [5, 3, 8, 4, 2]

print("Original list:", numbers)
print("Ordered list:", bubble_sort(numbers))

```

```python
# Selection sorting Case

def selection_sort(list):
    n = len(list)

    for i in range(n):
        min = i

        for j in range(i + 1, n):
            if list[j] < list[min]:
                min = j

        list[i], list[min] = list[min], list[i]

    return list


numbers = [5, 3, 8, 4, 2]

print("Original list:", numbers)
print("Ordered list:", selection_sort(numbers))
```

```python
# Insertion sorting Case

def insertion_sort(list):
    for i in range(1, len(list)):
        actual = list[i]
        j = i - 1

        while j >= 0 and list[j] > actual:
            list[j + 1] = list[j]
            j -= 1

        list[j + 1] = actual

    return list


numbers = [5, 3, 8, 4, 2]

print("Original list:", numbers)
print("Ordered list:", insertion_sort(numbers))
```

```python
# Quicksort Case

def quick_sort(list):
    if len(list) <= 1:
        return list

    pivot = list[0]

    smaller = [x for x in list[1:] if x <= pivot]
    greater = [x for x in list[1:] if x > pivot]

    return quick_sort(smaller) + [pivot] + quick_sort(greater)


numbers = [5, 3, 8, 4, 2]

print("Original list:", numbers)
print("Ordered list:", quick_sort(numbers))
```
