```python
#1. Write a script to point out whether a number is odd or even.
number = int(input("Enter an integer number: "))
if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
```

```python
#2. Write a script that, given an integer number, returns the digit of the units. For instance, for 153 you should return 3.
number = int(input("Enter an integer number: "))
def get_units_digit(number):
    # Ensure the input is an integer
    if not isinstance(number, int):
        raise ValueError("Input must be an integer.")
    
    # Get the units digit using modulo operation
    units_digit = abs(number) % 10
    return units_digit
units_digit = get_units_digit(number)
print(f"The digit of the units in {number} is: {units_digit}")
```

```python
#3. Write a script that takes a number and represents that number with binary digits (do not use direct conversion from the language).
number = int(input("Enter an integer number to convert to binary: "))
def convert_to_binary(number):
    # Ensure the input is an integer
    if not isinstance(number, int):
        raise ValueError("Input must be an integer.")
    
    # Handle the case for zero
    if number == 0:
        return "0"
    
    binary_digits = []
    is_negative = number < 0
    number = abs(number)
    
    while number > 0:
        remainder = number % 2
        binary_digits.append(str(remainder))
        number //= 2
    
    # If the original number was negative, we can represent it in two's complement or just indicate it's negative
    if is_negative:
        return "-" + ''.join(reversed(binary_digits))
    
    return ''.join(reversed(binary_digits))

#4. Do the same thing with conversions to any other given base.
number = int(input("Enter an integer number to convert to a different base: "))
base = int(input("Enter the base (greater than or equal to 2): "))
def convert_to_base(number, base):
    # Ensure the input is an integer and the base is valid
    if not isinstance(number, int):
        raise ValueError("Input must be an integer.")
    if not isinstance(base, int) or base < 2:
        raise ValueError("Base must be an integer greater than or equal to 2.")
    
    # Handle the case for zero
    if number == 0:
        return "0"
    
    digits = []
    is_negative = number < 0
    number = abs(number)
    
    while number > 0:
        remainder = number % base
        digits.append(str(remainder))
        number //= base
    
    # If the original number was negative, we can represent it in two's complement or just indicate it's negative
    if is_negative:
        return "-" + ''.join(reversed(digits))
    
    return ''.join(reversed(digits))
```
