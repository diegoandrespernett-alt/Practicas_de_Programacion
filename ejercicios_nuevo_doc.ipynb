```python
#1. Write a script to point out whether a number is odd or even.
number = int(input("Enter an integer number: "))
if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
```

```python
#2. Write a script that, given an integer number, returns the digit of the units. For instance, for 153 you should return 3.
number = int(input("Enter an integer number: "))
def get_units_digit(number):
    # Ensure the input is an integer
    if not isinstance(number, int):
        raise ValueError("Input must be an integer.")
    
    # Get the units digit using modulo operation
    units_digit = abs(number) % 10
    return units_digit
units_digit = get_units_digit(number)
print(f"The digit of the units in {number} is: {units_digit}")
```

```python
#3. Write a script that takes a number and represents that number with binary digits (do not use direct conversion from the language).
number = int(input("Enter an integer number to convert to binary: "))
def convert_to_binary(number):
    # Ensure the input is an integer
    if not isinstance(number, int):
        raise ValueError("Input must be an integer.")
    
    # Handle the case for zero
    if number == 0:
        return "0"
    
    binary_digits = []
    is_negative = number < 0
    number = abs(number)
    
    while number > 0:
        remainder = number % 2
        binary_digits.append(str(remainder))
        number //= 2
    
    # If the original number was negative, we can represent it in two's complement or just indicate it's negative
    if is_negative:
        return "-" + ''.join(reversed(binary_digits))
    
    return ''.join(reversed(binary_digits))
print(f"The binary representation of {number} is: {convert_to_binary(number)}")
```

```python
#4. Do the same thing with conversions to any other given base.
number = int(input("Enter an integer number to convert to a different base: "))
base = int(input("Enter the base (greater than or equal to 2): "))
def convert_to_base(number, base):
    # Ensure the input is an integer and the base is valid
    if not isinstance(number, int):
        raise ValueError("Input must be an integer.")
    if not isinstance(base, int) or base < 2:
        raise ValueError("Base must be an integer greater than or equal to 2.")
    
    # Handle the case for zero
    if number == 0:
        return "0"
    
    digits = []
    is_negative = number < 0
    number = abs(number)
    
    while number > 0:
        remainder = number % base
        digits.append(str(remainder))
        number //= base
    
    # If the original number was negative, we can represent it in two's complement or just indicate it's negative
    if is_negative:
        return "-" + ''.join(reversed(digits))

    return ''.join(reversed(digits))
print(f"The representation of {number} in base {base} is: {convert_to_base(number, base)}")
```

```python
#5. Write a script to show all the even numbers between two numbers.
start = int(input("Enter the starting integer: "))
end = int(input("Enter the ending integer: "))
def show_even_numbers(start, end):
    # Ensure the inputs are integers
    if not isinstance(start, int) or not isinstance(end, int):
        raise ValueError("Both start and end must be integers.")
    
    # Ensure start is less than or equal to end
    if start > end:
        raise ValueError("Start must be less than or equal to end.")
    
    even_numbers = []
    for number in range(start, end + 1):
        if number % 2 == 0:
            even_numbers.append(number)
    
    return even_numbers
print(f"The even numbers between {start} and {end} are: {show_even_numbers(start, end)}")
```

```python
#6. Write a script to calculate all the divisors of a given number.
n = int(input("Enter an integer number to find its divisors: "))
def calculate_divisors(n):
    # Ensure the input is an integer
    if not isinstance(n, int):
        raise ValueError("Input must be an integer.")
    
    # Handle the case for zero
    if n == 0:
        return "All integers are divisors of 0."
    
    divisors = []
    for i in range(1, abs(n) + 1):
        if n % i == 0:
            divisors.append(i)
    
    return divisors
print(f"The divisors of {n} are: {calculate_divisors(n)}")
```

```python
#7. Write a script to calculate the greatest common divisor of two given numbers.
def calculate_gcd(a, b):
    # Ensure the inputs are integers
    if not isinstance(a, int) or not isinstance(b, int):
        raise ValueError("Both inputs must be integers.")

    # Handle the case for zero
    if a == 0:
        return b
    if b == 0:
        return a

    # Use Euclidean algorithm
    a, b = abs(a), abs(b)
    while b:
        a, b = b, a % b
    return a

# Get input from the user
num1 = int(input("Enter the first integer: "))
num2 = int(input("Enter the second integer: "))

# Calculate and display the GCD
gcd = calculate_gcd(num1, num2)
print(f"The greatest common divisor of {num1} and {num2} is: {gcd}")
```

```python
#8. Write a script to calculate the least common multiple of two given numbers.
num1 = int(input("Enter the first integer: "))
num2 = int(input("Enter the second integer: "))
def calculate_lcm(num1, num2):
    # Ensure the inputs are integers
    if not isinstance(num1, int) or not isinstance(num2, int):
        raise ValueError("Both inputs must be integers.")

    # Handle the case for zero
    if num1 == 0 or num2 == 0:
        return 0

    # Calculate LCM using the relationship between GCD and LCM
    def calculate_gcd(a, b):
        a, b = abs(a), abs(b)
        while b:
            a, b = b, a % b
        return a
    gcd = calculate_gcd(num1, num2)
    lcm = abs(num1 * num2) // gcd
    return lcm
print(f"The least common multiple of {num1} and {num2} is: {calculate_lcm(num1, num2)}")
```

```python
#9. Write a script to calculate the value of the following formula, provided the number n as an input: ∑(j=1..n) ∑(i=1..j) j · i2
n = int(input("Enter the value of n: "))
result = 0
for j in range(1, n + 1):
    for i in range(1, j + 1):
        result += j * i ** 2
print("The value of the formula is:", result)
```

```python
#10. Write a script to guess a certain number, stored into a variable at the beginning of the program. To do this, ask the user for a number N, and then point out either “greater” or “lesser” (depending on whether the given number it is greater or not as regards to N). The process ends when the user gets it right.
secret_number = 42
while True:
    guess = int(input("Guess the number: "))
    if guess > secret_number:
        print("Lesser")
    elif guess < secret_number:
        print("Greater")
    else:
        print("Correct!")
        break
```

```python
#11. Write a script to draw a text square of n elements on each side using * but not filling it.
n = int(input("Enter the value of n: "))
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("* ", end="")
        else:
            print("  ", end="")
    print()
```
