```python
#Ejercicio 1.10.4. Implementar algoritmos que permitan:
#a) Calcular el perímetro de un rectángulo dada su base y su altura.
base = float(input("Escriba la base del rectángulo aquí: "))
altura = float(input("Escriba la altura del rectángulo aquí: "))
perimetro = 2 * (base + altura)
print("El perímetro del rectángulo es " + str(perimetro))

#b) Calcular el área de un rectángulo dada su base y su altura.
area = base * altura
print("El área del rectángulo es " + str(area))

#c) Calcular el área de un rectángulo (alineado con los ejes 𝑥 e 𝑦) dadas sus coordenadas
#𝑥1, 𝑥2, 𝑦1, 𝑦2.
x1 = float(input("Escriba la coordenada x1 del rectángulo aquí: "))
x2 = float(input("Escriba la coordenada x2 del rectángulo aquí: "))
y1 = float(input("Escriba la coordenada y1 del rectángulo aquí: "))
y2 = float(input("Escriba la coordenada y2 del rectángulo aquí: "))
area_rectangulo = abs(x2 - x1) * abs(y2 - y1)
print("El área del rectángulo es " + str(area_rectangulo))
#d) Calcular el perímetro de un círculo dado su radio.
radio = float(input("Escriba el radio del círculo aquí: "))
perimetro_circulo = 2 * 3.14159 * radio
print("El perímetro del círculo es " + str(perimetro_circulo))

#e) Calcular el área de un círculo dado su radio.
area_circulo = 3.14159 * radio ** 2
print("El área del círculo es " + str(area_circulo))

#f) Calcular el volumen de una esfera dado su radio.
volumen_esfera = (4/3) * 3.14159 * radio ** 3
print("El volumen de la esfera es " + str(volumen_esfera))

#g) Dados los catetos de un triángulo rectángulo, calcular su hipotenusa.
cateto1 = float(input("Escriba la longitud del primer cateto aquí: "))
cateto2 = float(input("Escriba la longitud del segundo cateto aquí: "))
hipotenusa = ((cateto1 ** 2 + cateto2 ** 2) ** 0.5)
print("La hipotenusa del triángulo es " + str(hipotenusa))
```

```python
#Ejercicio 1.10.6. Implementar un algoritmo que, dado un número entero 𝑛, permita calcular su
#factorial.
my_number = int(input("Escriba un número entero aquí: "))
factorial = 1
for i in range(1, my_number + 1):
    factorial *= i
print("El factorial de " + str(my_number) + " es " + str(factorial))
```

```python
#Ejercicio 1.10.7. Implementar algoritmos que resuelvan los siguientes problemas:
#a) Dados dos números, imprimir la suma, resta, división y multiplicación de ambos.
num1 = int(input("Escribe el primer numero: "))
num2 = int(input("Escribe el segundo numero: "))
suma = num1 + num2
resta = num1 - num2
division = num1 / num2
multiplicacion = num1 * num2
print("La suma de " + str(num1) + " y " + str(num2) + " es " + str(suma))
print("La resta de " + str(num1) + " y " + str(num2) + " es " + str(resta))
print("La división de " + str(num1) + " y " + str(num2) + " es " + str(division))
print("La multiplicación de " + str(num1) + " y " + str(num2) + " es " + str(multiplicacion))

#b) Dado un número entero 𝑛, imprimir su tabla de multiplicar.
n = int(input("Escribe un número entero: "))
i = 1
while i <= 12:
    print(i * n)
    i += 1
```
