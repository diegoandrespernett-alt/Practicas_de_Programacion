```python
#Ejercicio 5.7.1. Escribir un programa que permita al usuario ingresar un conjunto de notas, 
#preguntando a cada paso si desea ingresar más notas, e imprimiendo el promedio correspondiente.
notas = []
while True:
    nota = float(input("Ingresa una nota (o un número negativo para terminar): "))
    preguntar = input("¿Deseas ingresar otra nota? (s/n): ")
    if preguntar == "n":
        if nota >= 0:
            notas.append(nota)
        break
    else:
        if nota >= 0:
            notas.append(nota)
        continue

if len(notas) > 0:
    promedio = sum(notas) / len(notas)
    print("El promedio de las notas ingresadas es:", promedio)
```

```python
#Ejercicio 5.7.2. Escribir una función que reciba un número entero 𝑘 e imprima su descomposición en factores primos.
k = int(input("Ingresa un número entero mayor que 1 para descomponer en factores primos: "))
def descomponer_en_factores_primos(k):
    factores = []
    if k <= 1:
        print("El número debe ser mayor que 1.")
    if k > 1:
        divisor = 2
        while k > 1:
            while k % divisor == 0:
                factores.append(divisor)
                k //= divisor
            divisor += 1
    return factores

if k <= 1:
    print("El número debe ser mayor que 1.")
if k > 1:
    factores = descomponer_en_factores_primos(k)
    print("La descomposición en factores primos de", k, "es:", factores)
    
```

```python
#Ejercicio 5.7.3. Manejo de contraseñas
#a) Escribir un programa que contenga una contraseña inventada, que le pregunte al usuario la contraseña,
#  y no le permita continuar hasta que la haya ingresado correctamente.
contraseña_correcta = 301007
contraseña = int(input("Ingresa la contraseña: "))

while contraseña != contraseña_correcta:
    print("Contraseña incorrecta. Intenta de nuevo.")
    contraseña = int(input("Ingresa la contraseña: "))

if contraseña == contraseña_correcta:
    print("Contraseña correcta. ¡Bienvenido!")

#b) Modificar el programa anterior para que solamente permita una cantidad fija de intentos.
contraseña_correcta = 301007
intentos_maximos = 3
intentos = 0
contraseña = int(input("Ingresa la contraseña: "))

while contraseña != contraseña_correcta and intentos < intentos_maximos:
    print("Contraseña incorrecta. Intenta de nuevo.")
    contraseña = int(input("Ingresa la contraseña: "))
    intentos += 1

if contraseña == contraseña_correcta:
    print("Contraseña correcta. ¡Bienvenido!")
else:
    print("Has agotado el número de intentos.")

#c) Modificar el programa anterior para que después de cada intento agregue una pausa
#cada vez mayor, utilizando la función sleep del módulo time.
contraseña_correcta = 301007
intentos_maximos = 3
import time
contraseña = int(input("Ingresa la contraseña: "))

while contraseña != contraseña_correcta and intentos < intentos_maximos:
    print("Contraseña incorrecta. Intenta de nuevo.")
    print("Espera " + str(2 ** intentos) + " segundos.")
    contraseña = int(input("Ingresa la contraseña: "))
    intentos += 1

if contraseña == contraseña_correcta:
    print("Contraseña correcta. ¡Bienvenido!")
else:
    print("Has agotado el número de intentos.")

#d) Modificar el programa anterior para que sea una función que devuelva si el usuario
#ingresó o no la contraseña correctamente, mediante un valor booleano (True o False).
def verificar_contraseña(contraseña_correcta, intentos_maximos):
    import time
    intentos = 0
    contraseña = int(input("Ingresa la contraseña: "))

    while contraseña != contraseña_correcta and intentos < intentos_maximos:
        print("Contraseña incorrecta. Intenta de nuevo.")
        print("Espera " + str(2 ** intentos) + " segundos.")
        time.sleep(2 ** intentos)
        contraseña = int(input("Ingresa la contraseña: "))
        intentos += 1

    if contraseña == contraseña_correcta:
        print("Contraseña correcta. ¡Bienvenido!")
        return True
    else:
        print("Has agotado el número de intentos.")
        return False
```
